```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

print("Entrenando la red neuronal...")

# 1. Datos (y = 2x - 1)
x = np.array([-1.0, 0.0, 1.0, 2.0, 3.0, 4.0], dtype=float)
y = np.array([-3.0, -1.0, 1.0, 3.0, 5.0, 7.0], dtype=float)

# 2. Modelo
modelo = tf.keras.Sequential([tf.keras.layers.Dense(units=1, input_shape=[1])])
modelo.compile(optimizer='adam', loss='mean_squared_error')

# 3. Entrenamiento (guardamos el historial)
historial = modelo.fit(x, y, epochs=100, verbose=0)

# 4. Visualización
plt.figure(figsize=(12, 4))

# Gráfico 1: Cómo baja el error
plt.subplot(1, 2, 1)
plt.plot(historial.history['loss'], color='purple')
plt.title('Curva de Aprendizaje (Error)')
plt.xlabel('Intentos (Épocas)')
plt.ylabel('Margen de Error')

# Gráfico 2: Datos reales vs Predicción
plt.subplot(1, 2, 2)
plt.scatter(x, y, color='blue', label='Datos Reales')
predicciones = modelo.predict(x)
plt.plot(x, predicciones, color='red', label='Predicción IA')
plt.title('Ajuste de la Recta')
plt.legend()

# Mostrar los gráficos
plt.show()
```
